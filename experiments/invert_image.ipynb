```python
import cv2
import numpy as np
```

```python
# Read the image
img_path = "E:\\OneDrive - University of Moratuwa\\UoM-MSc-in-CS-2023\\MSc-Research\\References\\Images\\processed_acdc_structure.png"
image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Load in grayscale

# Identify distinct pixel values
unique_values = np.unique(image)
print("Distinct pixel values:", unique_values)
```

```python
# Invert the pixel values
inverted_image = 255 - image

# Save and display the result
img_out_path = img_path[:-4] + "_inverted.png"
cv2.imwrite(img_out_path, inverted_image)
cv2.imshow('Inverted Image', inverted_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Create a binary mapping (invert distinct values)
threshold = np.median(unique_values)  # Choosing median as a threshold
binary_image = np.where(image < threshold, 255, 0).astype(np.uint8)

cv2.imshow('Binary Inverted Image', binary_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
